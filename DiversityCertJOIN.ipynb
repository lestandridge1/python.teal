```python
import pandas as pd
import os
from datetime import datetime
```

```python
def find_input_file():
    # Get the current working directory
    directory = os.getcwd()
    
    # Look for the CSV file ending in "qualified Report"
    for filename in os.listdir(directory):
        if filename.endswith("qualified Report.csv"):
            return os.path.join(directory, filename)
    
    raise FileNotFoundError("No file ending in 'qualified Report.csv' found in the current directory.")
```

```python
def aggregate_certifications(input_file):
    # Load the input file with low_memory=False to avoid DtypeWarning
    df = pd.read_csv(input_file, low_memory=False)
    
    # Create a new column in column A titled "Diversity Certs"
    df.insert(0, 'Diversity Certs', '')
    
    # Identify all columns that end in "Certification"
    certification_columns = [col for col in df.columns if col.endswith('Certification')]
    
    # Populate the "Diversity Certs" column
    for index, row in df.iterrows():
        certifications = []
        for col in certification_columns:
            if row[col] == 'Yes':
                cert_name = col.replace(' Certification', '')
                certifications.append(cert_name)
        df.at[index, 'Diversity Certs'] = ', '.join(certifications)
    
    # Save the output file with the naming convention "Diversity Certs Aggregated - {today's date}"
    today = datetime.today().strftime('%Y-%m-%d')
    output_filename = f"Diversity Certs Aggregated - {today}.xlsx"
    df.to_excel(output_filename, index=False)
    print(f"Output file saved as {output_filename}")
```

```python
# Find the input file
input_file = find_input_file()

# Run the aggregation function
aggregate_certifications(input_file)
```
